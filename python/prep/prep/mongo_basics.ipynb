````python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client["Students"]

collection=db['studentscores']  
data = [ 
    {"user":"Krish", "subject":"Database", "score":80}, 
    {"user":"Amit",  "subject":"JavaScript", "score":90}, 
    {"user":"Amit",  "title":"Database", "score":85}, 
    {"user":"Krish",  "title":"JavaScript", "score":75}, 
    {"user":"Amit",  "title":"Data Science", "score":60},
    {"user":"Krish",  "title":"Data Science", "score":95}] 
  
# collection.insert_many(data) 
````

````python
for i in collection.find():
    print(i)
````

````python
### Find Amit And Krish Total Subjects
agg_result= collection.aggregate( 
    [{ 
    "$group" :  
        {"_id" : "$user",  
         "Total Subject" : {"$sum" : 1} 
         }}
    ]) 
for i in agg_result: 
    print(i)
````

````python
agg_result = collection.aggregate([{
    "$group":{
        "_id":"$user",
        "ts":{"$sum":1}
    }
}])
for i in agg_result:
    print(i)
````

````python
### Find Amit And Krish Total Marks

agg_result2 = collection.aggregate(
  [
      {
          "$group":
          {
              "_id":"$user",
              "Total marks":{"$sum":"$score"}
          }
      }
  ]  
)

for i in agg_result2: 
    print(i)
````

````python
### Calculating the average score based on user

agg_result3 = collection.aggregate(
  [
      {
          "$group":
          {
              "_id":"$user",
              "Total marks":{"$sum":"$score"},
              "average score":{"$avg":"$score"}
          }
      }
  ]  
)

for i in agg_result3: 
    print(i)
````

GitHub Copilot: The `$project` stage in MongoDB's aggregation framework is used to select specific fields from the documents in the pipeline, effectively changing the structure of the documents.

You can use `$project` to:

- Include specific fields from the original documents (by setting them to 1 or true).
- Exclude specific fields from the original documents (by setting them to 0 or false).
- Include computed fields or sub-documents.
- Rename fields.

In your code:

```python
for row in collection.aggregate( [ { "$project": { "title": 1,"isbn":1 } } ] ):
    print(row)
```

The `$project` stage is including only the `title` and `isbn` fields from each document in the `collection`. All other fields are excluded. The resulting documents will only contain these two fields (plus the `_id` field, which is included by default unless explicitly excluded).

````python
for row in collection.aggregate( [ { "$project": { "title": 1,"isbn":1 } } ] ):
    print(row)
````
